```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

## Load the data

```python
df.info()
df.describe()
```

## Plot: revenue by month

```python
plt.figure(figsize=(8, 4))
plt.plot(df["month"], df["revenue"], marker="o")
plt.title("Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.show()
```
